### Common types of plots
1. Line 
2. Scatter
3. Bar
4. Histogram
5. Subplots()

```python
import numpy as np
```

```python
np.random.random(size=(2,4)) # gives random normally distributed random number
```

```python
evenly_random = np.linspace(0,10,100) # np.linspace generates structured, evenly spaced numbers over a specified mathematical interval
evenly_random[:10]
```

## Line plot

```python
fig,ax = plt.subplots()
ax.plot(evenly_random,np.square(evenly_random));
```

# Scatter Plot

```python
fig,ax = plt.subplots(figsize=(10,10))
ax.scatter(x=evenly_random,y=np.exp(evenly_random))
plt.show()
```

```python
fig,ax = plt.subplots(figsize=(10,10))

ax.scatter(x=evenly_random,y=np.sin(evenly_random))

plt.show()
```

# Bar plot
- Horizontal Plot
- Vertical Plot

```python
import pandas as pd
fruit_prices = pd.DataFrame({'Apple':10,'Mango':90,'Banana':40},index=[0])
fruit_prices
```

```python
fruit_prices.columns
```

```python
fig,ax = plt.subplots()

ax.bar(x=fruit_prices.columns,height=fruit_prices.iloc[0]);
```

```python
fruits_dict = {'Apple':10,'Mango':90,'Banana':40}
fig,ax = plt.subplots()

ax.barh(y=fruits_dict.keys(),width=fruits_dict.values());
```

# Histogram

```python
normal_dist = np.random.randn(100) # random numbers with the given size from a noraml distribution
normal_dist[:10]
```

```python
fig,ax = plt.subplots()

ax.hist(x=normal_dist);
```

```python
random_num = np.random.random(size=100)
random_num[:10]
```

```python
fig,ax=plt.subplots()
ax.hist(x=random_num);
```

```python
np.random.rand(100)[0]
```

- np.random.rand -> gives float random number
- np.random.randn -> gives random number from a noraml dist
- np.random.randint -> gives int random number between given range

## Subplots

```python
fig,((ax1,ax2),(ax3,ax4)) = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax1.plot(evenly_random,np.square(evenly_random))
ax1.set(title='Line plot')

ax2.bar(x=fruits_dict.keys(),height=fruits_dict.values())
ax2.set(title='Bar chart')

ax3.scatter(x=evenly_random,y=np.sin(evenly_random))
ax3.set(title='Scatter')

ax4.hist(x=normal_dist)
ax4.set(title='Histogram')

fig.suptitle('Subplots demo')

plt.show();
```

## Another way of ploting subplots

```python
fig,ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].plot(evenly_random,np.square(evenly_random))
ax[0,0].set(title='Line plot')

ax[0,1].bar(x=fruits_dict.keys(),height=fruits_dict.values())
ax[0,1].set(title='Bar chart')

ax[1,0].scatter(x=evenly_random,y=np.sin(evenly_random))
ax[1,0].set(title='Scatter')

ax[1,1].hist(x=normal_dist)
ax[1,1].set(title='Histogram')

fig.suptitle('Subplots demo')

plt.show();
```
