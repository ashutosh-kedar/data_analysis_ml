## Filling the missing values with the mean/avg. Not the best practice but ok for sometimes

```python
car_sales_missing['Odometer'].fillna(car_sales_missing['Odometer'].mean())
```

## this returns a new series not affecting the original df. Can use inplace=True to update the original DF

```python
car_sales_missing['Odometer'] = car_sales_missing['Odometer'].fillna(car_sales_missing['Odometer'].mean())
```

```python
car_sales = pd.read_csv('https://raw.githubusercontent.com/ashutosh-kedar/data_analysis_ml/refs/heads/main/data/car-sales.csv')
car_sales
```

```python
# When we use series as a new fields and the series len is less then the DF len, remaining get NaN
# if using python list it should be of same length
```

```python
fuel_per_km = [3,5,2,4,7,6,9,1,4,5]
car_sales['Fuel per km'] = fuel_per_km
car_sales
```

```python
car_sales.drop(columns=['Fuel per km'],inplace=True)
```

```python
car_sales
```

```python
fuel_per_km = [3,5,2,4,7,6,9,1,4,5]
car_sales['Fuel per 100 km'] = fuel_per_km
car_sales
```

```python
car_sales['Fuel Consumed'] = car_sales['Odometer (KM)']/100 * car_sales['Fuel per 100 km']
car_sales
```

```python
car_sales['Odometer (miles)'] = car_sales['Odometer (KM)'].map(lambda x: x/1.6)
car_sales
```
