Standard Deviation and Variance is the measure of spread of the data

The Higher Standard Deviation and Variance is higher is the data spread out from the mean.

Lower the Standard Deviation and Variance lower is the data spread of the data from the mean.

```python
low_spread = np.array([2,4,6,8,10,12,14])
high_spread = np.array([2,40,100,500,1000])
```

```python
low_spread
```

```python
high_spread
```

```python
np.std(low_spread)
```

```python
np.std(high_spread)
```

```python
np.var(low_spread),np.var(high_spread)
```

SD is sqrt of Var

```python
np.std(low_spread),np.sqrt(np.var(low_spread))
```

## Dot product with example

```python
np.random.seed(24)
sales_matrix = np.random.randint(30,size=(5,3))
sales_matrix
```

```python
import pandas as pd
sales_df = pd.DataFrame(data=sales_matrix,
                        columns = ['Mango','Apple','Banana'],
                        index=['Mon','Tues','Wed','Thur','Fri']
                        )
sales_df
```

```python
np.random.seed(24)
price_matrix = np.random.randint(low=5,high=10,size=(1,3))
price_matrix
```

```python
price_df = pd.DataFrame(data=price_matrix,columns=['Mango','Apple','Banana'])
price_df
```

```python
price_matrix.shape
```

```python
sales_matrix.shape
```

```python
sales_matrix.T.shape
```

```python
daily_sales = np.dot(price_matrix,sales_matrix.T)
daily_sales
```

```python
daily_sales.shape
```

```python
sales_df
```

```python
sales_df['Daily Sales'] = daily_sales.T
sales_df
```

```python
price_df
```
